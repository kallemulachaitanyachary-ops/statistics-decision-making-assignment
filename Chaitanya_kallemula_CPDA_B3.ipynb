# Question 1: One-Sample Hypothesis Test (Altona Property Price)

## Objective
To test whether the average property price in Altona is significantly greater than $800,000.

### Hypotheses

Null Hypothesis (H0):
μ = 800000

Alternative Hypothesis (H1):
μ > 800000

### Significance Level
α = 0.05

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# Load dataset
df = pd.read_csv("property.csv")

# Display first 5 rows
df.head()
```

```python
# Filter data for Altona suburb
altona_data = df[df['Suburb'] == 'Altona']

# Extract price column and remove missing values
altona_prices = altona_data['Price'].dropna()

# Display sample size
print("Number of properties in Altona:", len(altona_prices))
```

```python
# Hypothesized population mean
population_mean = 800000

# Perform one-sample t-test
t_statistic, p_value_two_tailed = stats.ttest_1samp(altona_prices, population_mean)

# Since this is a right-tailed test:
p_value_one_tailed = p_value_two_tailed / 2

# Sample mean
sample_mean = altona_prices.mean()

# Print results
print("Sample Mean:", round(sample_mean, 2))
print("T-Statistic:", round(t_statistic, 4))
print("One-Tailed P-Value:", round(p_value_one_tailed, 4))
print("Significance Level (alpha): 0.05")
```

```python
alpha = 0.05

if (p_value_one_tailed < alpha) and (sample_mean > population_mean):
    print("Reject the Null Hypothesis")
    print("Conclusion: Average property price in Altona has significantly increased above $800,000.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Conclusion: There is not enough evidence to say that the average property price has increased above $800,000.")
```

## Interpretation

A one-sample t-test was conducted to determine whether the average property price in Altona is greater than $800,000.

Based on the p-value obtained from the test:

- If p-value < 0.05 → Reject H0
- If p-value > 0.05 → Fail to Reject H0

The test helps determine whether the increase in average property prices is statistically significant.

# Question 2: Summer vs Winter Property Prices (2016)

## Objective
To determine whether there is a significant difference in property prices between summer and winter months during 2016.

### Seasonal Classification
- Winter Months: October to March
- Summer Months: April to September

### Hypotheses

Null Hypothesis (H0):
There is no significant difference in mean property prices between summer and winter months.

Alternative Hypothesis (H1):
There is a significant difference in mean property prices between summer and winter months.

### Significance Level
α = 0.05

```python
# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Extract year and month
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Display first few rows
df[['Date', 'Year', 'Month']].head()
```

```python
# Filter dataset for year 2016
data_2016 = df[df['Year'] == 2016]

print("Total records in 2016:", len(data_2016))
```

```python
# Define winter months (October to March)
winter_months = [10, 11, 12, 1, 2, 3]

# Separate winter and summer data
winter_prices = data_2016[data_2016['Month'].isin(winter_months)]['Price'].dropna()

summer_prices = data_2016[~data_2016['Month'].isin(winter_months)]['Price'].dropna()

# Display sample sizes
print("Winter Properties:", len(winter_prices))
print("Summer Properties:", len(summer_prices))
```

```python
# Perform independent two-sample t-test
t_statistic, p_value = stats.ttest_ind(winter_prices, summer_prices, equal_var=False)

# Print results
print("Winter Mean Price:", round(winter_prices.mean(), 2))
print("Summer Mean Price:", round(summer_prices.mean(), 2))
print("T-Statistic:", round(t_statistic, 4))
print("P-Value:", round(p_value, 4))
print("Significance Level (alpha): 0.05")
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("Conclusion: There is a significant difference in property prices between summer and winter months.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Conclusion: There is no significant difference in property prices between summer and winter months.")
```

## Interpretation

An independent two-sample t-test was conducted to compare property prices between summer and winter months in 2016.

Decision Rule:
- If p-value < 0.05 → Reject H0
- If p-value > 0.05 → Fail to Reject H0

The test determines whether seasonal differences significantly affect property prices.

# Question 3: Probability of Properties Without Car Parking

## Objective
To calculate the probability that out of 10 properties sold in Abbotsford, exactly 3 properties will not have car parking space.

### Statistical Method
Binomial Probability Distribution

### Formula
P(X = x) = nCx * (p^x) * (1-p)^(n-x)

Where:
- n = total trials
- x = desired outcomes
- p = probability of success

```python
from scipy.stats import binom
```

```python
# Filter Abbotsford data
abbotsford_data = df[df['Suburb'] == 'Abbotsford']

# Display first few rows
abbotsford_data.head()
```

```python
# Count properties with no car parking
no_parking = abbotsford_data[abbotsford_data['Car'] == 0]

# Probability of no parking
p = len(no_parking) / len(abbotsford_data)

print("Probability of No Parking:", round(p, 3))
```

```python
# Given values
n = 10
x = 3

# Binomial probability
probability = binom.pmf(x, n, p)

print("Binomial Probability:", round(probability, 3))
```

## Interpretation

The binomial probability distribution was used to calculate the probability of exactly 3 properties without car parking out of 10 property sales in Abbotsford.

The calculated probability represents the likelihood of observing exactly 3 such properties in 10 independent sales.

# Question 6: One-Sample Hypothesis Test (Richmond Property Pricing)

## Objective
To determine whether the average property price in Richmond is significantly different from $1,000,000.

### Hypotheses

Null Hypothesis (H0):
μ = 1000000

Alternative Hypothesis (H1):
μ ≠ 1000000

### Significance Level
α = 0.05

```python
# Filter Richmond suburb data
richmond_data = df[df['Suburb'] == 'Richmond']

# Extract price column and remove missing values
richmond_prices = richmond_data['Price'].dropna()

# Display sample size
print("Number of properties in Richmond:", len(richmond_prices))
```

```python
# Hypothesized population mean
population_mean = 1000000

# Perform one-sample t-test
t_statistic, p_value = stats.ttest_1samp(richmond_prices, population_mean)

# Calculate sample mean
sample_mean = richmond_prices.mean()

# Display results
print("Sample Mean:", round(sample_mean, 2))
print("T-Statistic:", round(t_statistic, 4))
print("P-Value:", round(p_value, 4))
print("Significance Level (alpha): 0.05")
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("Conclusion: The average property price in Richmond is significantly different from $1,000,000.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Conclusion: There is not enough evidence to say that the average property price in Richmond differs from $1,000,000.")
```

## Business Conclusion

A one-sample t-test was conducted to evaluate the real estate firm's pricing claim for Richmond properties.

Decision Rule:
- If p-value < 0.05 → Reject H0
- If p-value > 0.05 → Fail to Reject H0

The test helps determine whether the claimed average property price aligns with actual market conditions.

# Question 7: Independent Two-Sample T-Test (Impact of Car Parking on Property Prices)

## Objective
To determine whether properties with car parking sell at a higher average price than properties without car parking.

### Choice of Test
Independent Two-Sample T-Test

### Reason for Choosing the Test
This test is appropriate because:
- There are two independent groups
- The dependent variable (Price) is continuous
- We are comparing average prices between groups

### Hypotheses

Null Hypothesis (H0):
Properties with car parking do not sell at higher average prices.

Alternative Hypothesis (H1):
Properties with car parking sell at higher average prices.

### Significance Level
α = 0.05

```python
# Properties with parking
parking_prices = df[df['Car'] > 0]['Price'].dropna()

# Properties without parking
no_parking_prices = df[df['Car'] == 0]['Price'].dropna()

# Display sample sizes
print("Properties with parking:", len(parking_prices))
print("Properties without parking:", len(no_parking_prices))
```

```python
# Perform independent two-sample t-test
t_statistic, p_value_two_tailed = stats.ttest_ind(
    parking_prices,
    no_parking_prices,
    equal_var=False
)

# Convert to one-tailed p-value
p_value_one_tailed = p_value_two_tailed / 2

# Calculate mean prices
parking_mean = parking_prices.mean()
no_parking_mean = no_parking_prices.mean()

# Display results
print("Mean Price (With Parking):", round(parking_mean, 2))
print("Mean Price (Without Parking):", round(no_parking_mean, 2))
print("T-Statistic:", round(t_statistic, 4))
print("One-Tailed P-Value:", round(p_value_one_tailed, 4))
print("Significance Level (alpha): 0.05")
```

```python
alpha = 0.05

if (p_value_one_tailed < alpha) and (parking_mean > no_parking_mean):
    print("Reject the Null Hypothesis")
    print("Conclusion: Properties with parking sell at significantly higher average prices.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Conclusion: There is not enough evidence to conclude that properties with parking sell at higher prices.")
```

## Interpretation of p-value

The p-value indicates whether the observed difference in average property prices is statistically significant.

Decision Rule:
- If p-value < 0.05 → Reject H0
- If p-value > 0.05 → Fail to Reject H0

## Business Implications for Developers

If properties with parking spaces command significantly higher prices, developers may prioritize parking infrastructure in future residential projects.

Otherwise, parking may not significantly influence selling prices based on this dataset.

# Question 8: Two-Way ANOVA (Suburb and Property Type)

## Objective
To determine whether property prices are influenced by:
- Suburb
- Type of Property
- Interaction between suburb and property type

### Statistical Technique
Two-Way ANOVA

### Variables

Dependent Variable:
- Price

Independent Variables:
- Suburb
- Type of Property

### Significance Level
α = 0.05

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
# Select a few suburbs for analysis
selected_suburbs = ['Richmond', 'Abbotsford', 'Altona']

anova_data = df[df['Suburb'].isin(selected_suburbs)]

# Keep required columns only
anova_data = anova_data[['Suburb', 'Type', 'Price']].dropna()

# Display first few rows
anova_data.head()
```

```python
# Build ANOVA model
model = ols(
    'Price ~ C(Suburb) + C(Type) + C(Suburb):C(Type)',
    data=anova_data
).fit()

# Generate ANOVA table
anova_table = sm.stats.anova_lm(model, typ=2)

# Display ANOVA results
anova_table
```

```python
print("Interpretation of ANOVA Results:\n")

alpha = 0.05

for index, row in anova_table.iterrows():
    
    # Skip residual row
    if index == 'Residual':
        continue
    
    p_value = row['PR(>F)']
    
    print(f"{index}")
    print(f"P-Value: {round(p_value, 4)}")
    
    if p_value < alpha:
        print("Result: Statistically Significant\n")
    else:
        print("Result: Not Statistically Significant\n")
```

## Interpretation

Two-Way ANOVA was conducted to examine whether property prices are significantly affected by:
- Suburb
- Property Type
- Interaction between Suburb and Property Type

Decision Rule:
- If p-value < 0.05 → Factor significantly affects property prices
- If p-value > 0.05 → Factor does not significantly affect property prices

The interaction effect helps determine whether the influence of property type changes across suburbs.

# Question 10: Industry-Style Hypothesis Validation (Bathroom Premium)

## Objective
To determine whether properties with more than 2 bathrooms command significantly higher prices.

### Statistical Test
Independent Two-Sample T-Test

### Reason for Choosing the Test
This test is suitable because:
- There are two independent groups
- Property price is a continuous variable
- The objective is to compare mean prices between groups

### Hypotheses

Null Hypothesis (H0):
Properties with more than 2 bathrooms do not command higher prices.

Alternative Hypothesis (H1):
Properties with more than 2 bathrooms command higher prices.

### Significance Level
α = 0.05

```python
# Properties with more than 2 bathrooms
more_than_2_bath = df[df['Bathroom'] > 2]['Price'].dropna()

# Properties with 2 or fewer bathrooms
two_or_less_bath = df[df['Bathroom'] <= 2]['Price'].dropna()

# Display sample sizes
print("Properties with >2 bathrooms:", len(more_than_2_bath))
print("Properties with <=2 bathrooms:", len(two_or_less_bath))
```

```python
# Perform independent t-test
t_statistic, p_value_two_tailed = stats.ttest_ind(
    more_than_2_bath,
    two_or_less_bath,
    equal_var=False
)

# Convert to one-tailed p-value
p_value_one_tailed = p_value_two_tailed / 2

# Calculate mean prices
mean_more_than_2 = more_than_2_bath.mean()
mean_two_or_less = two_or_less_bath.mean()

# Display results
print("Mean Price (>2 Bathrooms):", round(mean_more_than_2, 2))
print("Mean Price (<=2 Bathrooms):", round(mean_two_or_less, 2))
print("T-Statistic:", round(t_statistic, 4))
print("One-Tailed P-Value:", round(p_value_one_tailed, 4))
print("Significance Level (alpha): 0.05")
```

```python
alpha = 0.05

if (p_value_one_tailed < alpha) and (mean_more_than_2 > mean_two_or_less):
    print("Reject the Null Hypothesis")
    print("Conclusion: Properties with more than 2 bathrooms command significantly higher prices.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Conclusion: There is insufficient evidence to conclude that properties with more than 2 bathrooms command higher prices.")
```

## Policy Recommendation

The hypothesis test was conducted to evaluate whether properties with more than 2 bathrooms command premium prices.

Decision Rule:
- If p-value < 0.05 → Reject H0
- If p-value > 0.05 → Fail to Reject H0

## Recommendation to Policymakers

If the test shows a statistically significant increase in prices for properties with more than 2 bathrooms, policymakers and developers may consider:
- Encouraging development of larger family homes
- Supporting premium housing infrastructure
- Targeting high-income buyer segments

Otherwise, bathroom count may not significantly influence property pricing decisions.
